## 2. Ejecutar selección de variables argumentada mediante filtros estadísticos

(matrices de

a partir de un modelo base. 

##### Matriz de Correlación

```python
import numpy as np

# 1. Calcular la matriz de correlación (por defecto usa Pearson)
correlation_matrix = X.corr()

# 2. Visualización con mapa de calor (Heatmap)
plt.figure(figsize=(20, 15))
sns.heatmap(
    correlation_matrix, 
    annot=False,      # Mantenemos 'False' para que no se superpongan los números
    cmap='coolwarm',  # Tonos fríos para correlación negativa, cálidos para positiva
    linewidths=0.5,
    center=0
)
plt.title('Matriz de Correlación de Características', fontsize=18)
plt.tight_layout()
plt.show()

# 3. Extracción de variables con alta multicolinealidad (Filtro estadístico)
umbral = 0.8  # Puedes ajustar este valor entre 0.7 y 0.9 según tu criterio
matriz_absoluta = correlation_matrix.abs()

# Desenrollar la matriz para facilitar la lectura de los pares
pares_correlacionados = matriz_absoluta.unstack().sort_values(ascending=False)

# Filtrar las auto-correlaciones (diagonal de 1.0) y aplicar el umbral
pares_filtrados = pares_correlacionados[(pares_correlacionados > umbral) & (pares_correlacionados < 1.0)]

print(f"\n--- Pares de variables con correlación absoluta mayor a {umbral} ---")
# Usamos drop_duplicates() porque la matriz es simétrica (A vs B es lo mismo que B vs A)
print(pares_filtrados.drop_duplicates())
```

##### Eliminación de Variables redundantes y filtro de clase intermedia
Se eliminan las variables de primer semestre por que el segundo semestre tiene una relación más inmediata con la actualidad

```python
%pip install scikit-learn
%pip install imbalanced-learn
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

# 1. Filtrar la clase 'Enrolled' y binarizar el Target
# Asumiendo que el target se llama 'Target' (ajusta el nombre si es distinto)
mask = y['Target'] != 'Enrolled'
X_filtrado = X[mask].copy()
y_filtrado = y[mask]['Target'].map({'Dropout': 0, 'Graduate': 1}) # Binarizamos: 0 = Dropout, 1 = Graduate

# 2. Eliminar una variable de cada par altamente correlacionado
columnas_a_eliminar = [
    'Curricular units 1st sem (credited)',
    'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (grade)',
    "Father's occupation" # Altamente correlacionado con Mother's occupation
]
X_sin_colinealidad = X_filtrado.drop(columns=columnas_a_eliminar)

# 3. Estandarizar los datos (Lasso es sensible a la escala)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_sin_colinealidad)

# 4. Ajustar modelo con Regularización L1
# solver='liblinear' es necesario para L1 en sklearn
modelo_l1 = LogisticRegression(penalty='l1', solver='liblinear', random_state=42, C=0.1)
modelo_l1.fit(X_scaled, y_filtrado)

# 5. Visualizar los coeficientes resultantes
coeficientes = pd.Series(modelo_l1.coef_[0], index=X_sin_colinealidad.columns)
coeficientes_ordenados = coeficientes.abs().sort_values(ascending=False)

print("--- Importancia de Variables (Coeficientes L1) ---")
print(coeficientes_ordenados)

variables_descartadas = coeficientes_ordenados[coeficientes_ordenados == 0].index.tolist()
print(f"\nVariables con coeficiente 0 (puedes descartarlas): {variables_descartadas}")
```

```python
# Quitar las variables con coeficiente 0 para obtener el conjunto final de características
x_final = X_sin_colinealidad.drop(columns=variables_descartadas)
```

## 3. Modelo Dual

Para realizar la selección de modelos se tuvo en cuenta la cantidad de variables presentes en el dataset final, es por esto que se decidió hacer uso de la regresión logística (Baseline) y SVM (Retador).

### Datasets a usar: x_final y y_filtrado

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_final, y_filtrado, test_size=0.2, random_state=42, stratify=y_filtrado)
```

### Primer modelo: Regresión Logística

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import pandas as pd

# 1. Definir la estrategia de validación cruzada (K-Fold estratificado)
# Se usa StratifiedKFold para mantener la proporción de la variable objetivo
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Creación del Pipeline para evitar fuga de información (Data Leakage)
pipeline_baseline = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression(random_state=42, max_iter=1000))
])

# 3. Definición de la grilla de hiperparámetros para GridSearchCV
# Probamos penalización L1 (Lasso), L2 (Ridge) y ElasticNet con diferentes valores de C
param_grid_baseline = [
    {
        'logistic__penalty': ['l2'],
        'logistic__C': [0.001, 0.01, 0.1, 1, 10, 100],
        'logistic__solver': ['lbfgs', 'saga']
    },
    {
        'logistic__penalty': ['l1'],
        'logistic__C': [0.001, 0.01, 0.1, 1, 10, 100],
        'logistic__solver': ['saga']
    },
    {
        'logistic__penalty': ['elasticnet'],
        'logistic__C': [0.01, 0.1, 1, 10],
        'logistic__l1_ratio': [0.2, 0.5, 0.7],
        'logistic__solver': ['saga']
    }
]

# 4. Configuración del GridSearchCV
grid_baseline = GridSearchCV(
    estimator=pipeline_baseline,
    param_grid=param_grid_baseline,
    cv=cv_strategy,
    scoring='f1_weighted',  #Se usa F1_weighted para balancear la importancia de las clases en caso de desbalance
    n_jobs=-1,
    return_train_score=True
)

# 5. Ajustar el modelo
grid_baseline.fit(x_train, y_train)

# 6. Mostrar mejores hiperparámetros y resultado inicial
print("Mejores hiperparámetros Baseline:", grid_baseline.best_params_)
print("Mejor puntuación en CV (f1_weighted):", grid_baseline.best_score_)

import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# 2. SELECCIÓN DEL MEJOR ESTIMADOR
# Extraemos el modelo óptimo re-entrenado automáticamente
best_logistic_model = grid_baseline.best_estimator_

# 3. GENERACIÓN DE PREDICCIONES
# Predecimos únicamente sobre el conjunto de prueba reservado (x_test)
y_pred_baseline = best_logistic_model.predict(x_test)

# 4. REPORTE DE EVALUACIÓN Y MÉTRICAS
print("\n--- Reporte de Clasificación sobre x_test (Baseline) ---")
print(classification_report(y_test, y_pred_baseline))

# 5. MATRIZ DE CONFUSIÓN
cm_baseline = confusion_matrix(y_test, y_pred_baseline)
disp_baseline = ConfusionMatrixDisplay(
    confusion_matrix=cm_baseline, 
    display_labels=best_logistic_model.classes_
)

fig, ax = plt.subplots(figsize=(6, 5))
disp_baseline.plot(cmap='Blues', ax=ax)
plt.title("Matriz de Confusión - Regresión Logística (Baseline) en Test")
plt.show()
```

### Segundo modelo: SVM

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 1. Estrategia de Validación Cruzada (5-Fold Estratificado)
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Pipeline (Estandarización + SVM) para evitar fuga de datos
pipeline_retador = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(random_state=42))
])

# 3. Grilla de Hiperparámetros para SVM (Evaluando C, kernel y gamma)
param_grid_retador = [
    {
        'svm__kernel': ['linear'],
        'svm__C': [0.01, 0.1, 1, 10, 100]
    },
    {
        'svm__kernel': ['rbf'],
        'svm__C': [0.01, 0.1, 1, 10, 100],
        'svm__gamma': ['scale', 'auto', 0.01, 0.1, 1]
    },
    {
        'svm__kernel': ['poly'],
        'svm__degree': [2, 3],
        'svm__C': [0.1, 1, 10]
    }
]

# 4. Configuración de GridSearchCV
grid_retador = GridSearchCV(
    estimator=pipeline_retador,
    param_grid=param_grid_retador,
    cv=cv_strategy,
    scoring='f1_weighted',  # Métrica adecuada para desbalance de clases
    n_jobs=-1,
    return_train_score=True
)

# 5. Entrenar el modelo con el conjunto de entrenamiento
grid_retador.fit(x_train, y_train)

print("Mejores hiperparámetros Retador (SVM):", grid_retador.best_params_)
print("Mejor puntuación F1-Weighted en CV:", grid_retador.best_score_)

best_svm_model = grid_retador.best_estimator_

# 3. GENERACIÓN DE PREDICCIONES
# Se generan las predicciones finales sobre el conjunto de prueba reservado (X_test)
y_pred_retador = best_svm_model.predict(x_test)

# 4. EVALUACIÓN Y REPORTE DE MÉTRICAS
print("\n--- Reporte de Clasificación sobre X_test ---")
print(classification_report(y_test, y_pred_retador))

# 5. MATRIZ DE CONFUSIÓN
cm = confusion_matrix(y_test, y_pred_retador)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=best_svm_model.classes_)

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(cmap='Blues', ax=ax)
plt.title("Matriz de Confusión - Modelo Retador (SVM) en Test")
plt.show()
```
