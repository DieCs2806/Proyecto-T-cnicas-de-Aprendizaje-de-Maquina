##### Matriz de Correlación

```python
import numpy as np

# 1. Calcular la matriz de correlación (por defecto usa Pearson)
# Variables realmente numéricas (continuas o conteos)
cols_numericas = [
    'Previous qualification (grade)', 'Admission grade', 'Age at enrollment',
    'Curricular units 1st sem (credited)', 'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (evaluations)', 'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (grade)', 'Curricular units 1st sem (without evaluations)',
    'Curricular units 2nd sem (credited)', 'Curricular units 2nd sem (enrolled)',
    'Curricular units 2nd sem (evaluations)', 'Curricular units 2nd sem (approved)',
    'Curricular units 2nd sem (grade)', 'Curricular units 2nd sem (without evaluations)',
    'Unemployment rate', 'Inflation rate', 'GDP'
]

X_num = X[cols_numericas]
correlation_matrix = X_num.corr()

# 2. Visualización con mapa de calor (Heatmap)
plt.figure(figsize=(20, 15))
sns.heatmap(
    correlation_matrix, 
    annot=False,      # Mantenemos 'False' para que no se superpongan los números
    cmap='coolwarm',  # Tonos fríos para correlación negativa, cálidos para positiva
    linewidths=0.5,
    center=0
)
plt.title('Matriz de Correlación de Características', fontsize=18)
plt.tight_layout()
plt.show()

# 3. Extracción de variables con alta multicolinealidad (Filtro estadístico)
umbral = 0.8  # Puedes ajustar este valor entre 0.7 y 0.9 según tu criterio
matriz_absoluta = correlation_matrix.abs()

# Desenrollar la matriz para facilitar la lectura de los pares
pares_correlacionados = matriz_absoluta.unstack().sort_values(ascending=False)

# Filtrar las auto-correlaciones (diagonal de 1.0) y aplicar el umbral
pares_filtrados = pares_correlacionados[(pares_correlacionados > umbral) & (pares_correlacionados < 1.0)]

print(f"\n--- Pares de variables con correlación absoluta mayor a {umbral} ---")
# Usamos drop_duplicates() porque la matriz es simétrica (A vs B es lo mismo que B vs A)
print(pares_filtrados.drop_duplicates())
```

##### Eliminación de Variables redundantes y filtro de clase intermedia
Se eliminan las variables de primer semestre por que el segundo semestre tiene una relación más inmediata con la actualidad

```python
%pip install scikit-learn
%pip install imbalanced-learn
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# 1. Filtrar la clase 'Enrolled' y binarizar el Target
# Asumiendo que el target se llama 'Target' (ajusta el nombre si es distinto)
# 1. Binarizar el Target: 1 = Dropout, 0 = No deserta (Enrolled + Graduate)
X_filtrado = X.copy()
y_filtrado = (y['Target'] == 'Dropout').astype(int) # Binarizamos: 0 = Dropout, 1 = Graduate

# 2. Eliminar una variable de cada par altamente correlacionado
columnas_a_eliminar = [
    'Curricular units 1st sem (credited)',
    'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (grade)',
    #Se agregan las siguientes variables ya que solo entregan sesgo a la clasificacion, teniendo la gran mayoria de sus respuestas en una sola respuesta
    'Nacionality',
    'Educational special needs'
]
X_sin_colinealidad = X_filtrado.drop(columns=columnas_a_eliminar)

#Se agregan las siguientes variables ya que solo entregan sesgo a la clasificacion, teniendo la gran mayoria de sus respuestas en una sola respuesta


cols_categoricas = ['Marital Status', 'Application mode', 'Course', 'Previous qualification', 
                    "Mother's qualification", "Father's qualification",
                    "Mother's occupation"]
X_sin_colinealidad = pd.get_dummies(X_sin_colinealidad, columns=cols_categoricas, drop_first=True)

# 3. SPLIT AQUÍ, antes de cualquier ajuste
x_train, x_test, y_train, y_test = train_test_split(
    X_sin_colinealidad, y_filtrado, test_size=0.2, random_state=42, stratify=y_filtrado
)

# 4. Scaler y Lasso ajustados SOLO con train
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(x_train)
modelo_l1 = LogisticRegression(penalty='l1', solver='liblinear', random_state=42, C=0.1)
modelo_l1.fit(X_train_scaled, y_train)

# 5. Coeficientes (usar x_train.columns)
coeficientes = pd.Series(modelo_l1.coef_[0], index=x_train.columns)
coeficientes_ordenados = coeficientes.abs().sort_values(ascending=False)
variables_descartadas = coeficientes_ordenados[coeficientes_ordenados == 0].index.tolist()


print("--- Importancia de Variables (Coeficientes L1) ---")
print(coeficientes_ordenados)

variables_descartadas = coeficientes_ordenados[coeficientes_ordenados == 0].index.tolist()

print(f"\nVariables con coeficiente 0 (puedes descartarlas): {variables_descartadas}")
```

```python
# Quitar las variables con coeficiente 0 para obtener el conjunto final de características
x_train = x_train.drop(columns=variables_descartadas)
x_test =  x_test.drop(columns=variables_descartadas)
print("--- Información de las Variables Predictoras (X) ---")
display(x_train.info())
```

### Primer modelo: Regresión Logística

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import pandas as pd

# 1. Definir la estrategia de validación cruzada (K-Fold estratificado)
# Se usa StratifiedKFold para mantener la proporción de la variable objetivo
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Creación del Pipeline para evitar fuga de información (Data Leakage)
pipeline_baseline = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression(random_state=42, max_iter=1000, class_weight='balanced'))
])

# 3. Definición de la grilla de hiperparámetros para GridSearchCV
# Probamos penalización L1 (Lasso), L2 (Ridge) y ElasticNet con diferentes valores de C
param_grid_baseline = [
    {
        'logistic__penalty': ['l2'],
        'logistic__C': [0.001, 0.01, 0.1, 1, 10, 100],
        'logistic__solver': ['lbfgs', 'saga']
    },
    {
        'logistic__penalty': ['l1'],
        'logistic__C': [0.001, 0.01, 0.1, 1, 10, 100],
        'logistic__solver': ['saga']
    },
    {
        'logistic__penalty': ['elasticnet'],
        'logistic__C': [0.01, 0.1, 1, 10],
        'logistic__l1_ratio': [0.2, 0.5, 0.7],
        'logistic__solver': ['saga']
    }
]

# 4. Configuración del GridSearchCV
grid_baseline = GridSearchCV(
    estimator=pipeline_baseline,
    param_grid=param_grid_baseline,
    cv=cv_strategy,
    scoring='f1_weighted',  #Se usa F1_weighted para balancear la importancia de las clases en caso de desbalance
    n_jobs=-1,
    return_train_score=True
)

# 5. Ajustar el modelo
grid_baseline.fit(x_train, y_train)

# 6. Mostrar mejores hiperparámetros y resultado inicial
print("Mejores hiperparámetros Baseline:", grid_baseline.best_params_)
print("Mejor puntuación en CV (f1_weighted):", grid_baseline.best_score_)

import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# 2. SELECCIÓN DEL MEJOR ESTIMADOR
# Extraemos el modelo óptimo re-entrenado automáticamente
best_logistic_model = grid_baseline.best_estimator_

# 3. GENERACIÓN DE PREDICCIONES
# Predecimos únicamente sobre el conjunto de prueba reservado (x_test)
y_pred_baseline = best_logistic_model.predict(x_test)

# 4. REPORTE DE EVALUACIÓN Y MÉTRICAS
print("\n--- Reporte de Clasificación sobre x_test (Baseline) ---")
print(classification_report(y_test, y_pred_baseline))

# 5. MATRIZ DE CONFUSIÓN
cm_baseline = confusion_matrix(y_test, y_pred_baseline)
disp_baseline = ConfusionMatrixDisplay(
    confusion_matrix=cm_baseline, 
    display_labels=best_logistic_model.classes_
)

fig, ax = plt.subplots(figsize=(6, 5))
disp_baseline.plot(cmap='Blues', ax=ax)
plt.title("Matriz de Confusión - Regresión Logística (Baseline) en Test")
plt.show()
```

### Segundo modelo: SVM

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 1. Estrategia de Validación Cruzada (5-Fold Estratificado)
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Pipeline (Estandarización + SVM) para evitar fuga de datos
pipeline_retador = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(random_state=42,class_weight='balanced'))
])

# 3. Grilla de Hiperparámetros para SVM (Evaluando C, kernel y gamma)
param_grid_retador = [
        {
        'svm__kernel': ['rbf'],
        'svm__C': [0.1, 1, 10, 100],
        'svm__gamma': ['scale', 0.001, 0.01, 0.1]
    },
    {
        'svm__kernel': ['poly'],
        'svm__degree': [2, 3],
        'svm__C': [0.1, 1, 10],
        'svm__gamma': ['scale', 0.01, 0.1],
        'svm__coef0': [0, 1]
    }
]

# 4. Configuración de GridSearchCV
grid_retador = GridSearchCV(
    estimator=pipeline_retador,
    param_grid=param_grid_retador,
    cv=cv_strategy,
    scoring='f1_weighted',  # Métrica adecuada para desbalance de clases
    n_jobs=-1,
    return_train_score=True
)

# 5. Entrenar el modelo con el conjunto de entrenamiento
grid_retador.fit(x_train, y_train)

print("Mejores hiperparámetros Retador (SVM):", grid_retador.best_params_)
print("Mejor puntuación F1-Weighted en CV:", grid_retador.best_score_)

best_svm_model = grid_retador.best_estimator_

# 3. GENERACIÓN DE PREDICCIONES
# Se generan las predicciones finales sobre el conjunto de prueba reservado (X_test)
y_pred_retador = best_svm_model.predict(x_test)

# 4. EVALUACIÓN Y REPORTE DE MÉTRICAS
print("\n--- Reporte de Clasificación sobre X_test ---")
print(classification_report(y_test, y_pred_retador))

# 5. MATRIZ DE CONFUSIÓN
cm = confusion_matrix(y_test, y_pred_retador)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=best_svm_model.classes_)

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(cmap='Blues', ax=ax)
plt.title("Matriz de Confusión - Modelo Retador (SVM) en Test")
plt.show()
```

```python
# ============================================================
# PUNTO 4b - Curvas de Validación y Curvas de Aprendizaje
# Pegar después de las celdas de GridSearchCV (usa: x_train, y_train,
# grid_baseline, grid_retador, best_logistic_model, best_svm_model, cv_strategy)
# ============================================================

# %% [CELDA 1] Imports y funciones auxiliares
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.base import clone
from sklearn.model_selection import validation_curve, learning_curve

SCORING = 'f1_weighted'   # misma métrica que usaron en el GridSearchCV


def plot_validation_curve(estimator, X, y, param_name, param_range, titulo,
                          xlabel, log_x=True, ax=None):
    """Curva de validación: score de train vs CV al variar UN hiperparámetro."""
    train_scores, cv_scores = validation_curve(
        clone(estimator), X, y,
        param_name=param_name, param_range=param_range,
        cv=cv_strategy, scoring=SCORING, n_jobs=-1
    )
    tr_m, tr_s = train_scores.mean(axis=1), train_scores.std(axis=1)
    cv_m, cv_s = cv_scores.mean(axis=1), cv_scores.std(axis=1)

    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))
    plot = ax.semilogx if log_x else ax.plot
    plot(param_range, tr_m, 'o-', label='Entrenamiento', color='tab:blue')
    plot(param_range, cv_m, 's-', label='Validación cruzada', color='tab:orange')
    ax.fill_between(param_range, tr_m - tr_s, tr_m + tr_s, alpha=0.15, color='tab:blue')
    ax.fill_between(param_range, cv_m - cv_s, cv_m + cv_s, alpha=0.15, color='tab:orange')

    best_idx = cv_m.argmax()
    ax.axvline(param_range[best_idx], ls='--', color='gray',
               label=f'Mejor = {param_range[best_idx]:.3g}')
    ax.set_title(titulo)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(f'{SCORING}')
    ax.legend(loc='best')
    ax.grid(alpha=0.3)

    # Tabla resumen para argumentar en el informe (gap = indicador de varianza)
    resumen = pd.DataFrame({
        'valor': param_range, 'train': tr_m, 'cv': cv_m, 'gap (train-cv)': tr_m - cv_m
    }).round(4)
    return resumen


def plot_learning_curve(estimator, X, y, titulo, ax=None):
    """Curva de aprendizaje: score vs tamaño del conjunto de entrenamiento."""
    sizes, train_scores, cv_scores = learning_curve(
        clone(estimator), X, y,
        train_sizes=np.linspace(0.1, 1.0, 8),
        cv=cv_strategy, scoring=SCORING, n_jobs=-1, shuffle=True, random_state=42
    )
    tr_m, tr_s = train_scores.mean(axis=1), train_scores.std(axis=1)
    cv_m, cv_s = cv_scores.mean(axis=1), cv_scores.std(axis=1)

    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))
    ax.plot(sizes, tr_m, 'o-', label='Entrenamiento', color='tab:blue')
    ax.plot(sizes, cv_m, 's-', label='Validación cruzada', color='tab:orange')
    ax.fill_between(sizes, tr_m - tr_s, tr_m + tr_s, alpha=0.15, color='tab:blue')
    ax.fill_between(sizes, cv_m - cv_s, cv_m + cv_s, alpha=0.15, color='tab:orange')
    ax.set_title(titulo)
    ax.set_xlabel('Tamaño del conjunto de entrenamiento')
    ax.set_ylabel(f'{SCORING}')
    ax.legend(loc='best')
    ax.grid(alpha=0.3)
    print(f"{titulo}\n  Gap final (train - cv): {tr_m[-1] - cv_m[-1]:.4f} | "
          f"CV final: {cv_m[-1]:.4f} | Train final: {tr_m[-1]:.4f}")


# %% [CELDA 2] Curva de validación - REGRESIÓN LOGÍSTICA (Baseline): parámetro C
# C es el inverso de la regularización: C pequeño = más regularización = más sesgo, menos varianza
rango_C = np.logspace(-3, 2, 11)
res_lr_C = plot_validation_curve(
    best_logistic_model, x_train, y_train,
    param_name='logistic__C', param_range=rango_C,
    titulo='Curva de validación - Regresión Logística (C)',
    xlabel='C (inverso de la regularización)'
)
plt.show()
display(res_lr_C)

# %% [CELDA 3] Curva de validación - REGRESIÓN LOGÍSTICA: l1_ratio (mezcla L1/L2)
# Solo aplica si el mejor modelo usó penalty='elasticnet'
if best_logistic_model.get_params().get('logistic__penalty') == 'elasticnet':
    rango_l1 = np.linspace(0.0, 1.0, 9)
    res_lr_l1 = plot_validation_curve(
        best_logistic_model, x_train, y_train,
        param_name='logistic__l1_ratio', param_range=rango_l1,
        titulo='Curva de validación - Regresión Logística (l1_ratio)',
        xlabel='l1_ratio (0 = Ridge, 1 = Lasso)', log_x=False
    )
    plt.show()
    display(res_lr_l1)

# %% [CELDA 4] Preparar el SVM con kernel RBF (el verdadero "retador" no lineal)
# En su GridSearch ganó kernel='linear'; para analizar la parte no lineal se extrae
# la mejor combinación RBF de cv_results_ y se estudian C y gamma sobre ella.
cvr = pd.DataFrame(grid_retador.cv_results_)
mejor_rbf = cvr[cvr['param_svm__kernel'] == 'rbf'].sort_values('mean_test_score', ascending=False).iloc[0]
C_rbf, gamma_rbf = mejor_rbf['param_svm__C'], mejor_rbf['param_svm__gamma']
print(f"Mejor SVM-RBF en el grid: C={C_rbf}, gamma={gamma_rbf}, "
      f"CV={mejor_rbf['mean_test_score']:.4f}")

svm_rbf = clone(best_svm_model).set_params(
    svm__kernel='rbf', svm__C=C_rbf, svm__gamma=gamma_rbf
)
# Si gamma quedó como 'scale'/'auto' (string), se fija en un valor numérico para poder barrer
if isinstance(gamma_rbf, str):
    n_feat = x_train.shape[1]
    gamma_num = 1.0 / n_feat   # con datos estandarizados, 'scale' ~ 1/n_features
    svm_rbf.set_params(svm__gamma=gamma_num)

# %% [CELDA 5] Curva de validación - SVM: parámetro C (margen)
rango_C_svm = np.logspace(-2, 2, 9)
res_svm_C = plot_validation_curve(
    svm_rbf, x_train, y_train,
    param_name='svm__C', param_range=rango_C_svm,
    titulo='Curva de validación - SVM RBF (C)',
    xlabel='C (penalización por error / margen)'
)
plt.show()
display(res_svm_C)

# %% [CELDA 6] Curva de validación - SVM: parámetro gamma (alcance del kernel RBF)
rango_gamma = np.logspace(-4, 0, 9)
res_svm_gamma = plot_validation_curve(
    svm_rbf, x_train, y_train,
    param_name='svm__gamma', param_range=rango_gamma,
    titulo='Curva de validación - SVM RBF (gamma)',
    xlabel='gamma (alcance de cada muestra)'
)
plt.show()
display(res_svm_gamma)

# %% [CELDA 7] Curvas de aprendizaje de ambos modelos (lado a lado)
fig, axes = plt.subplots(1, 3, figsize=(20, 5), sharey=True)
plot_learning_curve(best_logistic_model, x_train, y_train,
                    'Aprendizaje - Regresión Logística (Baseline)', ax=axes[0])
plot_learning_curve(best_svm_model, x_train, y_train,
                    f'Aprendizaje - SVM mejor del grid ({best_svm_model.get_params()["svm__kernel"]})', ax=axes[1])
plot_learning_curve(svm_rbf, x_train, y_train,
                    'Aprendizaje - SVM RBF (retador no lineal)', ax=axes[2])
plt.tight_layout()
plt.show()
```
